```python
from typing import Literal
import requests


class _GPTOssResponse_Output_Content:
    def __init__(
        self,
        annotations: list | None = None,
        text: str | None = None,
        type: Literal["output_text", "reasoning_text"] | None = None,
        logprobs: list | None = None,
    ):
        self.annotations = annotations
        self.text = text
        self.type = type
        self.logprobs = logprobs


class _GPTOssResponse_OutPut:
    def __init__(
        self,
        id: str | None = None,
        content: list | None = None,
        summary: list | None = None,
        role: Literal["user", "assistant", "system"] | None = None,
        type: Literal["message", "reasoning_text"] | None = None,
        encrypted_content: str | None = None,
        status: str | None = None,
    ):
        self.id = id
        self.content = [
            _GPTOssResponse_Output_Content(**item) if isinstance(item, dict) else item
            for item in (content or [])
        ]
        self.summary = summary
        self.role = role
        self.type = type
        self.encrypted_content = encrypted_content
        self.status = status


class _GPTOssResponse_Usage:
    def __init__(
        self,
        input_tokens: int | None = None,
        output_tokens: int | None = None,
        total_tokens: int | None = None,
    ):
        self.input_tokens = input_tokens
        self.output_tokens = output_tokens
        self.total_tokens = total_tokens


class GPTOssResponse:
    def __init__(
        self,
        id: str | None = None,
        created_at: int | None = None,
        instructions: list | None = None,
        metadata: dict | None = None,
        model: str | None = None,
        object: Literal["response", "error"] | None = None,
        output: list | None = None,
        parallel_tool_calls: bool | None = None,
        temperature: float | None = None,
        tool_choice: str | None = None,
        tools: list | None = None,
        top_p: float | None = None,
        background: bool | None = None,
        max_output_tokens: int | None = None,
        max_tool_calls: int | None = None,
        previous_response_id: str | None = None,
        prompt: list | None = None,
        reasoning: list | None = None,
        service_tier: str | None = None,
        status: Literal["completed", "failed", "pending"] | None = None,
        text: str | None = None,
        top_logprobs: int | None = None,
        truncation: Literal["disabled", "enabled"] | None = None,
        usage: dict | None = None,
        user: str | None = None,
    ):
        self.id = id
        self.created_at = created_at
        self.instructions = instructions
        self.metadata = metadata
        self.model = model
        self.object = object
        self.output = [
            _GPTOssResponse_OutPut(**item) if isinstance(item, dict) else item
            for item in (output or [])
        ]
        self.parallel_tool_calls = parallel_tool_calls
        self.temperature = temperature
        self.tool_choice = tool_choice
        self.tools = tools
        self.top_p = top_p
        self.background = background
        self.max_output_tokens = max_output_tokens
        self.max_tool_calls = max_tool_calls
        self.previous_response_id = previous_response_id
        self.prompt = prompt
        self.reasoning = reasoning
        self.service_tier = service_tier
        self.status = status
        self.text = text
        self.top_logprobs = top_logprobs
        self.truncation = truncation
        self.usage = (
            _GPTOssResponse_Usage(**usage) if isinstance(usage, dict) else usage
        )
        self.user = user
```

```python
messages = [
    {
        "role": "system",
        "content": "Bạn là một trợ lý hữu ích. Ngôn ngữ sử dụng là tiếng Việt.",
    },
    {"role": "user", "content": "Cho tôi thời gian chi tiết hiện tại ở Việt Nam."},
]
```

```python
response = requests.post(
    "https://ai.maiquyen.name.vn/api/chat",
    json={
        # "model": "@openai/gpt-oss-120b",
        # "lora-adapter": "@maiquyen/cdn/lora-adapter/gpt-oss-120b-vietnamese-lora-adapter",
        "messages": messages + [{"role": "user", "content": "Tôi vừa hỏi gì?"}],
        "temperature": 0.7,
        "top_p": 0.9,
    },
)
```

```python
resp = GPTOssResponse(**response.json())
assistant_text = (
    resp.output[0].content[0].text if resp.output and resp.output[0].content else None
)
if assistant_text:
    messages.append({"role": "assistant", "content": assistant_text})
messages
```
